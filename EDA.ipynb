```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid")
```

```python
df = pd.read_csv('../data/AmesHousing.csv.xls')
```

## Univariate Analysis

### Histogram with KDE - SalePrice (numerical)

```python
sns.histplot(df['SalePrice'], kde=True)
plt.show()
```

### Boxplot - SalePrice (numerical, check distribution and outliers)

```python
sns.boxplot(x=df['SalePrice'])
plt.show()
```

### Countplot - House Style (categorical)

```python
sns.countplot(x='House Style', data=df)
plt.xticks(rotation=45)
plt.show()
```

## Bivariate Analysis

### Lineplot - Year Built vs SalePrice (numerical vs numerical)

```python
sns.lineplot(x='Year Built', y='SalePrice', data=df)
plt.show()
```

### Scatterplot - Gr Liv Area vs SalePrice (numerical vs numerical)

```python
sns.scatterplot(x='Gr Liv Area', y='SalePrice', data=df)
plt.show()
```

### Boxplot - Overall Qual vs SalePrice (categorical vs numerical)

```python
sns.boxplot(x='Overall Qual', y='SalePrice', data=df)
plt.show()
```

### Barplot - Neighborhood vs SalePrice (categorical vs numerical)

```python
plt.figure(figsize=(12, 6))
sns.barplot(x='Neighborhood', y='SalePrice', data=df)
plt.xticks(rotation=90)
plt.show()
```

## Multivariate Analysis

### Heatmap - Correlation of Numerical Columns

```python
plt.figure(figsize=(16, 12))
sns.heatmap(df.select_dtypes(include='number').corr(), cmap='coolwarm')
plt.show()
```
